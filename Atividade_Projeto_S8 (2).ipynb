```python
# Define a classe EcommerceSystem, responsável por gerenciar usuários, produtos e pedidos
class EcommerceSystem:

    # Método construtor da classe, executado ao criar um objeto
    def __init__(self):

        # Lista para armazenar os usuários cadastrados
        self.users = []

        # Lista para armazenar os produtos disponíveis
        self.products = []

        # Lista para armazenar os pedidos realizados
        self.orders = []

    # Método para adicionar um usuário ao sistema
    def add_user(self, user):

        # Adiciona o usuário na lista de usuários
        self.users.append(user)

    # Método para adicionar um produto ao sistema
    def add_product(self, product):

        # Adiciona o produto na lista de produtos
        self.products.append(product)

    # Método para realizar um pedido
    def place_order(self, user, product):

        # Verifica se o produto existe na lista de produtos
        if product in self.products:

            # Adiciona uma tupla (usuário, produto) na lista de pedidos
            self.orders.append((user, product))

            # Retorna mensagem de sucesso
            return "Order placed successfully"

        # Caso o produto não exista, retorna mensagem de erro
        return "Product not available"


# Exemplo de uso do sistema

# Cria uma instância da classe EcommerceSystem
system = EcommerceSystem()

# Adiciona um usuário chamado Alice
system.add_user("Alice")

# Adiciona um produto chamado Laptop
system.add_product("Laptop")

# Realiza um pedido e imprime o resultado na tela
print(system.place_order("Alice", "Laptop"))

```

Passo 2: Refatoração dde Microsserviços

```python

# Classe responsável por gerenciar usuários
class UserService:

    # Método construtor da classe
    def __init__(self):

        # Lista para armazenar os usuários cadastrados
        self.users = []

    # Método para adicionar um usuário
    def add_user(self, user):

        # Adiciona o usuário na lista de usuários
        self.users.append(user)


# Classe responsável por gerenciar produtos
class ProductService:

    # Método construtor da classe
    def __init__(self):

        # Lista para armazenar os produtos disponíveis
        self.products = []

    # Método para adicionar um produto
    def add_product(self, product):

        # Adiciona o produto na lista de produtos
        self.products.append(product)

    # Método para verificar se um produto está disponível
    def check_availability(self, product):

        # Retorna True se o produto existir na lista, caso contrário False
        return product in self.products


# Classe responsável por gerenciar pedidos
class OrderService:

    # Método construtor da classe
    def __init__(self):

        # Lista para armazenar os pedidos realizados
        self.orders = []

    # Método para realizar um pedido
    def place_order(self, user, product, product_service):

        # Verifica se o produto está disponível usando o serviço de produtos
        if product_service.check_availability(product):

            # Adiciona o pedido na lista de pedidos
            self.orders.append((user, product))

            # Retorna mensagem de sucesso
            return "Order placed successfully"

        # Retorna mensagem caso o produto não esteja disponível
        return "Product not available"


# Exemplo de uso utilizando microsserviços

# Cria uma instância do serviço de usuários
user_service = UserService()

# Cria uma instância do serviço de produtos
product_service = ProductService()

# Cria uma instância do serviço de pedidos
order_service = OrderService()

# Adiciona um usuário chamado Alice
user_service.add_user("Alice")

# Adiciona um produto chamado Laptop
product_service.add_product("Laptop")

# Realiza um pedido e imprime o resultado na tela
print(order_service.place_order("Alice", "Laptop", product_service))
```
